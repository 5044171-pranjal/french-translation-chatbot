```python
import os
from openai import OpenAI

# Initialize the OpenRouter/OpenAI client using the OPENROUTER_API_KEY environment variable
client = OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key=os.getenv("OPENROUTER_API_KEY")
)

print('OpenRouter client configured. Model will be: gemma-4-26b-a4b-it')
```

## 2. Model Initialization

We will use the `gemma-4-26b-a4b-it` model via the OpenRouter API. The client is initialized above; calls to the model occur in the translation function.

```python
# Model calls are made in the translation function below using `client`.
# No local model initialization is required when using OpenRouter-hosted models.
model_name = 'google/gemma-4-26b-a4b-it:free'
print('Configured to use model:', model_name)
```

## 3. Translation Function

Encapsulate the translation call in a small helper function that accepts English text and returns cleaned French text. This function centralizes model calls and any post-processing (e.g., trimming whitespace).

```python
def get_french_translation(text: str) -> str:
    """Translate English `text` to French using `gemma-4-26b-a4b-it` and return the result."""
    if not text or not text.strip():
        return ''
    system_prompt = (
        'You are a professional translator. Translate the users English text into natural, idiomatic French. ',
        'Respond ONLY with the French translation and nothing else (no commentary, no explanations).'
    )
    response = client.chat.completions.create(
        model='google/gemma-4-26b-a4b-it:free',
        messages=[
            {'role': 'system', 'content': ''.join(system_prompt)},
            {'role': 'user', 'content': text}
        ],
        max_tokens=400,
        temperature=0.0
    )
    return response.choices[0].message.content.strip()
```

```python
def type_out(text: str, delay: float = 0.02):
    """Print `text` one character at a time to simulate typing."""
    import sys, time
    for ch in text:
        sys.stdout.write(ch)
        sys.stdout.flush()
        time.sleep(delay)
    print()
```

```python
def run_translation_chatbot():
    print('🇬🇧→🇫🇷 English-to-French Translation Chatbot')
    print("Type an English sentence and press Enter. Type 'exit' or 'quit' to stop.")
    while True:
        try:
            text = input('You: ')
        except (KeyboardInterrupt, EOFError):
            print('\nExiting...')
            break
        if text.strip().lower() in ('exit', 'quit'):
            print('Goodbye!')
            break
        if not text.strip():
            continue
        try:
            french = get_french_translation(text)
            print('\nBot: ', end='', flush=True)
            type_out(french, delay=0.02)
            print()
        except Exception as e:
            print(f'Error during translation: {e}')
            print()

# Uncomment to run in a live session
# if __name__ == '__main__':
#     run_translation_chatbot()
```
